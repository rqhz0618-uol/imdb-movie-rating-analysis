## Aim

The aim of my M1 module was to extract structured answer sequences from raw quiz response files. Each respondent file contains 100 questions, and each question has four possible answers. My module converts each response into a list of 100 integers, where 1, 2, 3, and 4 represent selected answers, and 0 represents an unanswered question.

```python
from scripts.data_extraction_M1 import extract_answers_sequence, write_answers_sequence

file_path = 'data/answers_respondent_1.txt'
answers = extract_answers_sequence(file_path)

print('First 10 answers:', answers[:10])
print('Total answers:', len(answers))

write_answers_sequence(answers, 1, 'output')
print('Answer sequence written successfully.')
```

## Explanation and Testing

The function `extract_answers_sequence` reads a respondent text file line by line. It collects the four answer options for each question and checks which option contains `[x]`. If an option contains `[x]`, its position is recorded as the answer value. If no option is selected, the value 0 is recorded.

The function `write_answers_sequence` saves the extracted list into a text file using the required filename format `answers_list_respondent_n.txt`.

The module was tested using a respondent file. The output confirmed that the function returned 100 answers, which matches the number of questions in the quiz.
